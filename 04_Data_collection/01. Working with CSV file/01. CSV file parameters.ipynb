## 3 Opening a csv file from an url

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv" #paste url where u need to get csv file
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}  # keep hearder same 

req = requests.get(url, headers=headers)
data = StringIO(req.text)
df = pd.read_csv(data)
df.head(-20)
```

## 5. Index_col parameter

## 6. Header parameter

```python
pd.read_csv("eurusd_hour.csv", header=0 ) # specifing the row of header in case of any miss reading 
```

## 7. use_cols parameter

```python
pd.read_csv("eurusd_hour.csv", usecols=["Date", "BO", 'BH'])  # to specify necesary columns which dont read unnecessay columns
```

## 8. skiprows/nrow Parameter

```python
pd.read_csv("eurusd_hour.csv",skiprows=[1,3])
pd.read_csv("eurusd_hour.csv",nrows=100)
```

## 9. Encoding parameter

```python
pd.read_csv("eurusd_hour.csv",encoding="latin-1")  #to interpret the raw bytes in the file as text characters
```

```python
## Handeling date
```

```python
pd.read_csv("eurusd_hour.csv", parse_dates=["Date"])  # it helps to read date as date not as object(string)
```

```python
## converters
```

```python
def rename(Time):
    if Time == "01:00":
        return "1am"
    else: 
        return Time
```

```python
pd.read_csv("eurusd_hour.csv", converters={'Time':rename})  #custom function to transform the raw values of a specific column
```
