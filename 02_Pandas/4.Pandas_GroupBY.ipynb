# Group By Objects

```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv("movies.csv")
movies.head()
```

```python
ipl = pd.read_csv("ipl.csv")
ipl.head(2)
```

```python
genre = movies.groupby('Genre')
```

### Applying builtin aggregation functions on groupby objects 

```python
genre.sum().head()
```

```python
genre.min().head()
```

```python
# find the top 3 genres by total earning
```

```python
movies.groupby("Genre").sum()['Gross'].sort_values(ascending=False)
```

```python
movies.groupby("Genre")['Gross'].sum().sort_values(ascending=False)   # efficient compared to above 
```

```python
# find the genre with the highest avg IMDB rating
movies.groupby("Genre")['IMDB_Rating'].mean().sort_values(ascending = False).head(1)
```

```python
# find the director with most popular based on No_of_Votes
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending = False).head(1)
```

```python
# find the highest rates movie of each genre
movies.groupby('Genre')['IMDB_Rating'].max()
```

```python
# find number of movies done by actor 
# movies['Star1'].value_counts()

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending = False)
```

### Group by Attributes and Methods

```python
genre = movies.groupby("Genre")
```

```python
# find number of columns > len
len(genre)
```

```python
# total items in each group > size
genre.size()
```

```python
# first or last element of groups 
genre.first()
genre.last() # returns  last element of every groups

genre.nth(6)  # to get any index elements
```

```python
# get_group > vs filtering
genre.get_group("Horror")   # returns whole group of the category
```

```python
genre.groups
```

```python
# sample
genre.sample()  # reutruns one random movies from each groups
```

```python
# describe
genre.describe()
```

```python
# nunique
genre.nunique()
```

## agg method
help to perform different operations on different columns

```python
#passing dict
genre.agg(
    {

        "Runtime":"mean",
        "IMDB_Rating":"mean",
        "No_of_Votes":"sum"
    }
)  # perform mentioned opertions on mentioned columns 
```

```python
# using list 
genre.agg(['min','max'])   # performs mentiones operations to all columns
```

```python
# combined of both
genre.agg(
    {   
        "Runtime":"mean",
        "IMDB_Rating":["mean", 'max'],
        "No_of_Votes":"sum"
    }
)
```
