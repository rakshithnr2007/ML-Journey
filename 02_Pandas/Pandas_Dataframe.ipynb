###  Important DataFrame Functions

```python
# astype
ipl.info()  # initially 168.1KB
```

```python
ipl['season'] = ipl['season'].astype("int16")
ipl.info()  # later 164.7KB
```

```python
# by converting repeated values into categorical data
ipl['toss_decision'] = ipl['toss_decision'].astype("category")
ipl['team1'] = ipl['team1'].astype("category")
ipl['team2'] = ipl['team2'].astype("category")
ipl['result'] = ipl['result'].astype("category")
ipl['venue'] = ipl['venue'].astype("category")
ipl.info() # now 107.0 KB
```
