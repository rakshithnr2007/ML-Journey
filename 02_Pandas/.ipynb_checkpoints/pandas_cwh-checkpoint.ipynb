# Essential Pandas Concepts

## Missing Data & Cleaning
* **`isnull() / isna()`**: Detecting missing values
* **`dropna()`**: Dropping rows or columns with missing values
* **`fillna()`**: Replacing missing values with specific data
* **`duplicated()`**: Identifying duplicate rows in a dataset
* **`drop_duplicates()`**: Removing duplicate rows from a dataset

## Sorting & Structural Changes
* **`sort_values()`**: Sorting data by specific columns
* **`rename()`**: Changing column names or index labels
* **Creating new columns**: Deriving new data from existing columns

## Summaries & Descriptive Statistics
* **`describe()`**: Viewing a quick summary of statistics
* **`mean() / median() / min() / max() / sum() / std()`**: Calculating basic mathematical aggregations
* **`value_counts()`**: Counting the frequency of unique values
* **`unique() / nunique()`**: Finding unique entries and counting them
* **`corr()`**: Calculating statistical correlation between columns

## Grouping & Combining Data
* **`groupby()`**: Splitting data into distinct groups
* **`agg()`**: Applying multiple statistical functions to groups
* **`merge()` / Joins**: Combining different datasets using matching keys
* **`concat()`**: Stacking datasets together vertically or horizontally

## Strings & Dates
* **String `.str` operations**: Manipulating text data in columns
* **`pd.to_datetime()`**: Converting string text into actual date objects
* **`.dt` operations**: Extracting specific elements like day, month, or year
* **Time-series operations**: Grouping, shifting, or resampling date-based data

## Advanced Transformations
* **`pivot_table()`**: Creating spreadsheet-style summary tables
* **`apply() / map()`**: Running custom Python functions across rows or columns
