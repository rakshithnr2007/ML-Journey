## View vs Copy


```python
newdf[0][0]  # access a single value
```

```python
df1 = newdf  # this is a VIEW, not a copy
```

```python
newdf.loc[0, 0] = 65
newdf.head()
```

```python
df1.loc[0, 0] = 9  # changing df1 also changes newdf, since they share memory
df1.head()
```

### Making a Real Copy with `.copy()`
To avoid this shared-memory behavior, use `.copy()`. It creates a fully independent DataFrame — changes to the copy do **not** affect the original.

```python
df1 = df.copy()  # an independent copy
df1
```

```python
df1.loc[0, 'marks'] = 100  # only affects df1
df1
```

```python
df  # unchanged - proof that df1 is independent
```

## Dropping Rows and Columns


```python
newdf.drop(3, axis=0)  # drop row labeled 3
```

```python
newdf.drop(3, axis=1)  # drop column labeled 3
```

## Accessing Data — `loc` vs `iloc`


```python
newdf.loc[[1, 2], [0, 2]]  # rows 1 & 2, columns 0 & 2, by label
```

```python
newdf.loc[(newdf[0] < 0.5)]  # filter rows where column 0 is less than 0.5
```

```python
newdf.iloc[0, 1]  # access by position instead of label
```

## Saving Modifications Safely


```python
newdf.drop([0, 2], axis=1)  # NOT saved - just a preview, newdf itself is unchanged
newdf.head()
```

```python
newdf = newdf.drop([0], axis=1)  # saved, because we reassigned the result
newdf.head()
```

```python
# NOTE: this line has a common beginner bug!
# inplace=True already modifies newdf AND returns None.
# So reassigning it ("newdf = ...") overwrites newdf with None.
# Correct version would just be: newdf.drop([1], axis=1, inplace=True)
newdf = newdf.drop([1], axis=1, inplace=True)
```

## Resetting the Index


```python
df1 = pd.DataFrame(np.random.rand(200, 7), index=np.arange(200))
df1
```

```python
df1.drop(1, axis=1)  # preview only, not saved
```

```python
df1.reset_index(drop=True, inplace=True)  # roll back to a clean, sequential index
df1
```

```python
df1.loc[:, [3]] = 3  # set every value in column 3 to 3
df1
```
