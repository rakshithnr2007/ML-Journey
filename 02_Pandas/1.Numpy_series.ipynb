# Series
a Pandas series is like a column in a table . its a 1D array holding data of any type

### Series from lists

```python
# string 
players = ["virat","dhoni","dhavan ", "rohit"]

pd.Series(players)
```

```python
# integers
runs = [12,23,45,99]

pd.Series(runs)
```

```python
# custom index
marks = [67,56,45,89]
subject = ['maths','ss','kan','science']

pd.Series(marks , index=subject)
```

```python
#  setting a name

marks = [67,56,45,89]
subject = ['maths','ss','kan','science']

pd.Series(marks , index=subject, name="marks of test")
```

### Series from dict

```python
marks = {
    "maths" : 67,
    "social" : 33,
    'science': 49,
    'kan': 99
}
marks_series = pd.Series(marks , name="using dict ")
marks_series
```

### Series Attribute

```python
# size
marks_series.size
```

```python
# dtpe
marks_series.dtype
```

```python
# name
marks_series.name
```

```python
# is_unique  (returkns any existance of copy or not )
marks_series.is_unique
```

```python
# index    ( returns index value of series)
marks_series.index
```

```python
# values
marks_series.values
```
