# Types of Functions in Seaborn

### 1. Axes-level functions
These functions create a single plot on a specific Matplotlib Axes.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.title("Total Bill vs Tip")
plt.show()
```

### 2. Figure-level functions
These functions control the entire figure and can create multiple plots.

```python
import seaborn as sns

tips = sns.load_dataset("tips")  #tips is seaborn data set

sns.relplot(x="total_bill", y="tip", data=tips, kind="scatter" , height=6 ,aspect=2)

# we can adjust figure size in figure level by mentioning height and aspect 
```

# 1. Relational plots
Used to understand the relationship between two or more variables.
* bivariate analysis

## 1.1 Scatter plot

```python
# Scatter plot -> axes level function
sns.scatterplot(data=tips, x="total_bill", y='tip',hue='sex', style='time',size='size')  
# hue adds a categorical dimension using color
# style categorical variable using different marker shapes
# represent a variable by changing the size of the markers
```

```python
# relplot -> figure level plot
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex', style='time',size='size')
```

## 1.2 Line plot 

```python
gap = px.data.gapminder()
gap.head  # datasetm stored in plotly
```

```python
temp_df = gap[gap['country']=='India']
temp_df
```

```python
#axes level function
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
# figure level fucntion relplot
sns.relplot(data=temp_df, x='year', y='lifeExp' , kind='line' )
```

```python
# hue -> style
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df, x='year', y='lifeExp',hue='country',style='continent',size='iso_num')
```

### Facet plot
split one dataset into multiple plots based on categories.

```python
# facet plot  -> only wit fiufer level fuction -> relplot
sns.relplot(data=tips, x="total_bill", y='tip', kind='scatter', col='day', row='sex')
# creates plots side by side
# creates plots one below another
# hue separates categories using color
```

```python
# col wrap 
# to control how many facet plots can appear in one row.
sns.relplot(data=gap, x='lifeExp', y='gdpPercap',kind='scatter' , col='year', col_wrap=4)
```

# 2. Distribution Plots
Distribution plots are used to understand how the values of a variable are spread.
* used for univariate analysis
* used to find out the distribution
* range of the obsevation
* central tendency
* is the data bimodal? or are there outliers?

```python
# figure level fuction -> displot
# axes level function -> histplot , kdeplot , regplot
```

## 2.1 Hist plot

```python
# plotting univariate histogram
sns.histplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill',kind='hist')
```

```python
# bins parameter 
sns.displot(data=tips, x='total_bill',kind='hist', bins=15, hue='sex')  
#bins number of bins or groups
```

```python
# categorical data
# hist acting like count plot 
sns.displot(data=tips, x='day',kind='hist')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
#  facet
sns.displot(data=titanic, x='age',kind='hist', col='sex' , hue='survived' )
```

## 2.2 Kdeplot (Kernel Density Estimation) 
shape of a data distribution as a smooth curve.

```python
sns.kdeplot(data=tips , x='total_bill')
```

```python
sns.displot(kind='kde', data=tips, x='total_bill' , hue='sex', fill=True)
```

## 2.3 Rugplot
* individual data points along an axis using small lines called ticks.
( plot marginal distribution by drawing ticks along the x and y axis)
* this function is intended to complemet other plot by showing the locationof individual observvation

```python
sns.kdeplot(data=tips , x = 'total_bill')
sns.rugplot(data=tips , x = 'total_bill')
```

```python
# bivariate histogram 
# a bivariate histogram bins the data within rectanges that tile the plot 
# and then shows the count of observations within each rectangle with the fill color 

sns.histplot(data=tips , x='total_bill' , y='tip')
```

```python
# bivariate kde plot
# a bivariate KDE plt smoothes the (x,y) observations with a @d Gaussian
sns.displot(kind='kde', data=tips, x='total_bill', y='tip')
```

## 3. Matrix Plot
 used to visualize data that is arranged in a matrix/table format.
* Finding correlation between features
* Identifying patterns in numerical data
* Visualizing large tables of values

## 3.1 Heatmap 
represents values using different color-encoded matrix. 
* It is commonly used to visualize a correlation matrix.

```python
gap
```

```python
temp = gap.pivot(index='country' , columns='year' ,values='lifeExp')
temp
```

```python
# axes level function
plt.figure(figsize=(15,8))
sns.heatmap(temp)
```

```python
# annotation
temp = gap[gap["continent"]== 'Europe'].pivot(index='country' , columns='year' ,values='lifeExp')

plt.figure(figsize=(15,8))
sns.heatmap(temp , annot=True , linewidth=0.5 , cmap='jet')
```

## 3.2 Clustermap
matrix plot that creates a heatmap with hierarchical clustering.
    *It is useful when you want to see not only the values, but also which rows or columns are similar to each other.

```python
iris = px.data.iris()  # Iris flower measurements 
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
# algo used is agglomerative hierarcical clustering
```
