```python
import pandas as pd
import requests
```

```python
url = "https://api.tvmaze.com/shows"

response = requests.get(url)  # requesting data using url to api, in return we get data

response  # 200 got response , 400 not avaiable , 500 server down
```

```python
data = response.json() # it explicitly converts incoming data to json format

data
```

```python
df = pd.DataFrame(data)  # converting list of dictonaries to dataframe

df.head()
```

```python
df = df[['id', 'name', 'type', 'language', 'genres', 'status', 'runtime', 'rating', 'premiered']] 
#selecting required columns

df.head()
```

```python
# if having multiple pages 
```

```python

df = pd.DataFrame()

for page in range(5):
    url = f"https://api.tvmaze.com/shows?page={page}"  # looping replacing page number
    
    response = requests.get(url)
    data = response.json()
    
    temp_df = pd.DataFrame(data)
    
    df = pd.concat([df, temp_df], ignore_index=True)  # ignore index maintain continuty of serial 
```
