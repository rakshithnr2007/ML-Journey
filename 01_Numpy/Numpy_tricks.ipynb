## Set functions

```python
m = np.array([1,2,3,4,5,6,7])
n = np.array([1,3,5,7,9,11,13])
```

```python
np.union1d(m,n)
```

```python
np.intersect1d(m,n)
```

```python
np.setdiff1d(m,n)
```

```python
np.setxor1d(m,n)
```
