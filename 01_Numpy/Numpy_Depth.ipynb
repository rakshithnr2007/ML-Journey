# Broadcasting
Broadcasting in NumPy describes how the library treats arrays with different shapes during arithmetic operations. Subject to certain constraints, the smaller array is conceptually "broadcast" or stretched across the larger array so they have compatible shapes

```python
# same shape 
a = np.arange(12,24).reshape(3,4)
b = np.arange(24,36).reshape(3,4)

a+b
```

```python
# different shape
a = np.arange(3).reshape(1,3)
b = np.arange(18).reshape(2,3,3)

print(a)
print(b)
print(a + b )
```

* 1 (Same Dimensions): Turn the (3,) array into a (1, 3) array so both have 2 dimensions
* 2 (Match Sizes): Compare the sizes: (2, 3) vs (1, 3). The columns already match (3), and the rows are compatible because one of them is 1.
* 3 (Stretch & Execute): Replicate the row of the (1, 3) array 2 times to become (2, 3). NumPy then adds them together instantly.

### Mean square error
used to calculate loss function ( linear regresion)

```python
actual = np.random.randint(1,50,12).reshape(2,6)
predict = np.random.randint(1,50,12).reshape(2,6)
```

```python
print(actual)
print(predict)
```

```python
def mse(actual , predict):
    return np.mean((actual - predict)**2)

mse(actual , predict)
```
