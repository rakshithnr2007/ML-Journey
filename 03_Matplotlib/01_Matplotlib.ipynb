```python
# from a pandas dataframe
batsman = pd.read_csv("sharma-kohli.csv")
batsman.head()

plt.plot(batsman['index'], batsman['V Kohli'])
```

```python
# plotting multiple plots 
plt.plot(batsman['index'], batsman['V Kohli'])
plt.plot(batsman['index'], batsman['RG Sharma'])

plt.title('Rohit Sharma vs Virat Kohli carrrer comaprison')  # tilte labelling
plt.xlabel('season')  # x axis label
plt.ylabel('Runs scored')  # y axis label
```

```python
# colors(hex) ,
plt.plot(batsman['index'], batsman['V Kohli'],color='red')    # can mention name of color 
plt.plot(batsman['index'], batsman['RG Sharma'],color='#369FEB')  # can provide hex code 
```

```python
#  line(width and styke) 

plt.plot(batsman['index'], batsman['V Kohli'],linestyle='dashdot',linewidth=4)  # style ad width can be changed   
plt.plot(batsman['index'], batsman['RG Sharma'],linestyle='dashed',linewidth=3) 
#  dashdot , solid , none are other styles
```

```python
# marker(size)

plt.plot(batsman['index'], batsman['V Kohli'],linestyle='dashdot',linewidth=4, marker='D',markersize=7)   
plt.plot(batsman['index'], batsman['RG Sharma'],linestyle='dashed',linewidth=3, marker='o' )
#Symbol,Name
"""
o,Circle
s,Square
^,Triangle up
v,Triangle down
D,Diamond
*,Star
.,Point
+,Plus
x,X cross
p,Pentagon
"""
```

```python
# legend -> label  (to name the line )

plt.plot(batsman['index'], batsman['V Kohli'],color='red',linestyle='dashdot',marker='D', label="kohli")   
plt.plot(batsman['index'], batsman['RG Sharma'],color='blue',linestyle='solid',marker='D', label="Rohit")

plt.legend(loc='best') # this display labeling on graph and best parameter figure out best place to display
```

```python
# limiting axes
# when outlier gets in other graph becomes plain to keep the detail we do limit 
price = [48000,54000,57000,49000,47000,45000,500000] 
year  =  [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)    
plt.ylim(0,85000)
# plt.xlim(2015,2018)  x axis limit

# grid
plt.grid()
```

```python
# show  , this displays the plotted the graph 

plt.plot(batsman['index'], batsman['V Kohli'],color='red',linestyle='dashdot',marker='D', label="kohli")   
plt.plot(batsman['index'], batsman['RG Sharma'],color='blue',linestyle='solid',marker='D', label="Rohit")

plt.legend()

plt.show() # in notebook it calls byy itself other platform need to call manually
```

## Subplots 
To plot multiple graphs side by side

```python
# A diff way to plot graphs 
batters = pd.read_csv('batter.csv')
batters.head(100)
```

```python
# previous methods
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title='Titlee'
plt.xlabel('Average')
plt.ylabel('Strike_rate')
```

```python
#Second method 
fig, ax = plt.subplots(figsize=(15,6)) # figure  and axis objects formed

ax.scatter(batters['avg'],batters['strike_rate'], marker='+' ,color='red')
ax.set_title('Titleee')
ax.set_xlabel('Average')
ax.set_ylabel('Strike_rate')

```

```python
# 2 Subplots
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True, figsize=(12,6))  # figsize=(15,6),nrows=4,ncols=4 , it shares x axis coordinates

ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')  # ax[0] is first axis
ax[1].scatter(batters['avg'],batters['runs'],)

ax[0].set_title("Avg vs Strikerate")
ax[0].set_ylabel("Strikerate")

ax[1].set_title('Avg vs runs')
ax[1].set_ylabel("Runs")

ax[1].set_xlabel("Average")
```

```python
# 4 Subplots  > it stores and accessed by 2d array
fig, ax = plt.subplots(nrows=2,ncols=2)

ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red')  # ax[0] is first axis
ax[0,1].scatter(batters['avg'],batters['runs'],)
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['strike_rate'])
```

```python
# third method 
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)  # 2 rows 1 column  1 position (rows, columns, position)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.scatter(batters['avg'],batters['runs'],color='green')

ax3 = fig.add_subplot(2,2,3)
ax3.scatter(batters['strike_rate'],batters['runs'],color='green')
```

## 3D plots

### i. 3D Scatter plots

```python
batters = pd.read_csv("batter.csv")
```

```python
fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'])
ax.set_title("IPL baysman analysis")
ax.set_xlabel("runs")
ax.set_ylabel("average")
ax.set_zlabel("Strike Rate")
```

### ii. 3D line plot

```python
x = [0,1,4,3]
y = [0,10,15,12]
z = [0, 20,25 , 12]
fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z) # ploting points on x , y ,z axis

ax.plot3D(x,y,z , color="red")  # ploting or joining line between points
```

### iii. 3D Surface plots
Important for ploting loss funciton

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
x.shape , y.shape
```

```python
xx , yy = np.meshgrid(x,y)
# xx and yy are variables that store the 2D coordinate grids
```

```python
xx.shape , yy.shape   
```

```python
z = xx**2 + yy**2    # z = x^2 + y^2 , calculating all the possible combinations
z.shape
```

```python
fig = plt.figure(figsize=(15,8))

ax = plt.subplot(projection='3d')

g= ax.plot_surface(xx,yy,z, cmap='viridis') # viridis/jet is  perceptually uniform sequential colormap 
fig.colorbar(g)


ax.set_xlabel("x axis")
ax.set_ylabel("y axis")
ax.set_zlabel("z axis")
```

```python
z = np.sin(xx)-np.cos(yy)
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

g= ax.plot_surface(xx,yy,z, cmap='jet') 
fig.colorbar(g)
```

### iv. Contour plot 
reps 3D plot in 2D  
consider Top view

```python
z = xx**2 + yy**2    
fig = plt.figure(figsize=(15,8))

ax = plt.subplot()

g= ax.contour(xx,yy,z, cmap='jet')
fig.colorbar(g)
```

```python
# contourf or Filled contour graph
z = xx**2 + yy**2  

fig = plt.figure(figsize=(15, 8))
ax = plt.subplot()

g = ax.contourf(xx,yy,z, cmap='jet') # contourf f is added at end of function name
fig.colorbar(g)

plt.show()
```
